```python
import cv2

img = cv2.imread("/content/kevin.jpg")
```

```python
from google.colab.patches import cv2_imshow

cv2_imshow(img)
```

```python
img.shape
```

```python
img_redimensionada = cv2.resize(img, (384, 512))

cv2_imshow(img_redimensionada)
```

```python
gray_scale = cv2.cvtColor(img_redimensionada, cv2.COLOR_BGR2GRAY)

cv2_imshow(gray_scale)
```

```python
gray_scale.shape
```

```python
print(gray_scale)
```

```python
import numpy as np

media = 0
desvio_padrao = 50  # aumente para deixar o ruído mais visível

ruido = np.random.normal(
    loc=media,
    scale=desvio_padrao,
    size=gray_scale.shape
)

cv2_imshow(ruido)
```

```python
imagem_ruidosa = gray_scale.astype(np.float32) + ruido

# Garante valores válidos de pixel e converte de volta para imagem
imagem_ruidosa = np.clip(imagem_ruidosa, 0, 255).astype(np.uint8)
cv2_imshow(imagem_ruidosa)
```

```python
print('Média da Imagem Original: ', gray_scale.mean())
print('Média do Ruído: ', ruido.mean())
print('Média da Imagem Ruidosa: ', imagem_ruidosa.mean())
```
